```python
from langchain.tools import tool
from langchain.agents import create_agent
from IPython.display import Markdown, display, clear_output
import subprocess


@tool
def search(query: str) -> str:
    """Search for information."""
    return f"Results for: {query}"

@tool
def get_weather(location: str) -> str:
    """Get weather information for a location."""
    return f"Weather in {location}: Sunny, 72°F"

@tool
def get_ollama_models() -> str:
    """Get the current Ollama running models."""
    result = subprocess.run(["ollama", "ps"], capture_output=True, text=True)
    return result.stdout

@tool
def execute_command(command: str) -> str:
    """Execute a shell command and return the output."""
    result = subprocess.run(command, shell=True, capture_output=True, text=True)
    return result.stdout

agent = create_agent(llm, tools=[execute_command])

# response = agent.invoke({"messages": [("human", "Please run 'ollama list' and tell which models is best for reasoning tasks?")]})


stream = agent.stream({
    "messages": [("human", "Please run 'ollama list' and tell which models is best for reasoning tasks?")]
})

handle = display(Markdown(""), display_id=True)
full_text = ""

for step in stream:
    # step is a dict like {"messages": [...]}
    
    if "messages" in step:
        for msg in step["messages"]:
            if hasattr(msg, "content") and msg.content:
                full_text += msg.content
                handle.update(Markdown(full_text))
```

```python
from langchain.tools import tool
from langchain.agents import create_agent
import subprocess
from langchain_core.runnables import RunnableConfig
from langsmith import traceable

# Create InMemoryCheckpoint for agent state
from langgraph.checkpoint.memory import InMemorySaver

checkpoint = InMemorySaver()
config: RunnableConfig = {"configurable": {"thread_id": "1"}}
@tool
def execute_command(command: str) -> str:
    """Execute a shell command and return the output."""
    result = subprocess.run(command, shell=True, capture_output=True, text=True)
    return result.stdout

agent = create_agent(
    model=llm, 
    tools=[execute_command],
    system_prompt="You are an assistant that can execute shell commands to interact with the system. Use the 'execute_command' tool to run commands and provide the output in your response.",
    checkpointer=checkpoint
)

response = agent.invoke({"messages": [("human", "Please run 'ollama list' and tell which models is best for reasoning tasks?")]}, config=config)
# response = agent.invoke({"messages": [("human", "Which ollama models are running?")]}, config=config)
# response = agent.invoke({"messages": [("human", "create an text file named test.txt with content 'Hi there! how are you?' and then display the content of the file")]}, config=config)
print(response["messages"][-1].content)
```

```python
from dotenv import load_dotenv
load_dotenv('../../.env')

from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain.tools import tool
import subprocess
from langsmith import traceable


# Initialize the Ollama LLM
llm = ChatOllama(model="gemma4:e2b")

# Checkpointer for agent state
from langchain_core.runnables import RunnableConfig
from langgraph.checkpoint.memory import InMemorySaver

# Create InMemoryCheckpoint for agent state
checkpoint = InMemorySaver()
config: RunnableConfig = {"configurable": {"thread_id": "1"}}

# Create psql tool


@traceable
def psql(command: str, database: str = "postgres") -> str:
    """Execute a psql command against any database on the local Postgres instance.

    Set `database` to a specific database such as `emails`, `dvdrental`, or
    `accountdb`. Use `postgres` for server-level discovery commands like `\\l`.
    """
    psql_command = [
        "/opt/homebrew/opt/libpq/bin/psql",
        f"postgresql://postgres:postgres@localhost:5432/{database}",
        "-c",
        command,
    ]
    result = subprocess.run(psql_command, capture_output=True, text=True)
    if result.returncode != 0:
        return f"Error executing command: {result.stderr}"
    elif result.stdout.strip() == "":
        return "Command executed successfully, but no output was returned."
    return result.stdout


psql_tool = tool(psql)
agent = create_agent(
    model=llm,
    tools=[psql_tool],
    system_prompt="You are a helpful assistant that can execute psql commands to interact with a local Postgres instance. The 'psql' tool can access any local database by passing the `database` argument, such as `postgres`, `emails`, `dvdrental`, `accountdb`, `IPR_2`, `my_database`, `pgsql_cache`, or `pydantic_ai_sql_gen`. Use `database='postgres'` for discovery commands like `\\l`, and target a specific database for table inspection or SQL queries. Use the tool when needed and provide the results clearly in your response."
)



```

```python
from IPython.display import Markdown, display

resp = agent.invoke({"messages": [("human", "What is the version of PostgreSQL running on the system? and also tell me the current database name?")]})

display(Markdown(resp['messages'][-1].content))
```
