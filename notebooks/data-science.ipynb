```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    # "day": [1, 2, 3, 4, 5, 6],

    "temperature": [20, 22, np.nan, np.nan, np.nan, 0]

})

print(df)
```

```python
df['dropna_temp'] = df['temperature'].dropna()
df['bfill_temp'] = df['temperature'].bfill()
df['ffill_temp'] = df['temperature'].ffill()
df['interpolate_temp'] = df['temperature'].interpolate()

df
```
